## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Pada jam berapa rata-rata penyewaan sepeda paling tinggi, dan bagaimana perbedaannya antara hari kerja dan hari non-kerja selama periode 2011–2012?

- **Pertanyaan 2:** Bagaimana rata-rata jumlah penyewaan sepeda harian berbeda berdasarkan musim dan kondisi cuaca selama periode 2011–2012, dan kondisi mana yang perlu menjadi prioritas operasional?

## Exploratory Data Analysis (EDA)

### Explore ...

### Explore Pola Penyewaan Berdasarkan Jam dan Jenis Hari

Untuk menjawab pertanyaan bisnis pertama, dilakukan analisis rata-rata jumlah penyewaan sepeda berdasarkan jam dan jenis hari, yaitu hari kerja (`Workingday`) dan hari non-kerja (`Non-workingday`).

```python
# Menghitung rata-rata jumlah penyewaan berdasarkan jam dan jenis hari
hourly_rentals = (
    hour_df.groupby(["day_type", "hr"], as_index=False)["cnt"]
    .mean()
    .rename(columns={"cnt": "avg_rentals"})
)

# Mengurutkan hasil berdasarkan jenis hari dan jam
hourly_rentals = hourly_rentals.sort_values(
    by=["day_type", "hr"]
)

hourly_rentals.head(15)
```

```python
print(f"bentuk table hourly_rentals: {hourly_rentals.shape}")
```

```python
# Lima jam dengan rata-rata penyewaan tertinggi untuk setiap jenis hari
top_hours = (
    hourly_rentals
    .sort_values(["day_type", "avg_rentals"], ascending=[False, False])
    .groupby("day_type")
    .head(5)
)

top_hours
```

**Insight:** (Opsional)
- Pada hari kerja, rata-rata penyewaan tertinggi terjadi pada pukul 17.00, yaitu sekitar 525 sepeda, kemudian pukul 18.00 dengan rata-rata sekitar 492 sepeda.
- Pada hari kerja, penyewaan juga tinggi pada pukul 08.00 dengan rata-rata sekitar 477 sepeda. Hal ini menunjukkan adanya pola penyewaan pada jam berangkat dan pulang kerja.
- Pada hari non-kerja, puncak penyewaan bergeser ke siang hingga sore hari, terutama pada pukul 13.00 dengan rata-rata sekitar 373 sepeda.
- Pola ini menunjukkan bahwa pengelola perlu mempersiapkan ketersediaan sepeda lebih banyak pada pagi dan sore hari kerja, serta siang hingga sore hari non-kerja.

### Explore Pola Penyewaan Berdasarkan Musim dan Kondisi Cuaca

Untuk menjawab pertanyaan bisnis kedua, dilakukan analisis rata-rata jumlah penyewaan sepeda harian berdasarkan musim dan kondisi cuaca.

```python
# Menghitung rata-rata penyewaan berdasarkan musim
seasonal_rentals = (
    day_df.groupby("season_label", as_index=False)["cnt"]
    .mean()
    .rename(columns={"cnt": "avg_rentals"})
)

# Mengurutkan dari rata-rata tertinggi
seasonal_rentals = seasonal_rentals.sort_values(
    by="avg_rentals",
    ascending=False
)

seasonal_rentals
```

```python
# Menghitung rata-rata penyewaan berdasarkan kondisi cuaca
weather_rentals = (
    day_df.groupby("weather_label", as_index=False)["cnt"]
    .mean()
    .rename(columns={"cnt": "avg_rentals"})
)

# Mengurutkan dari rata-rata tertinggi
weather_rentals = weather_rentals.sort_values(
    by="avg_rentals",
    ascending=False
)

weather_rentals
```

**Insight:**

- Musim gugur (`Fall`) memiliki rata-rata penyewaan harian tertinggi, yaitu sekitar 5.644 sepeda.
- Musim semi (`Spring`) memiliki rata-rata penyewaan paling rendah, yaitu sekitar 2.604 sepeda.
- Pada kondisi cuaca cerah (`Clear`), rata-rata penyewaan mencapai sekitar 4.877 sepeda per hari.
- Ketika terjadi hujan atau salju ringan (`Light Rain/Snow`), rata-rata penyewaan turun menjadi sekitar 1.803 sepeda per hari.
- Kondisi cuaca buruk perlu menjadi prioritas dalam perencanaan operasional karena berkaitan dengan penurunan permintaan yang cukup besar.

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=hourly_rentals,
    x="hr",
    y="avg_rentals",
    hue="day_type",
    marker="o",
    linewidth=2.5
)

plt.title(
    "Rata-rata Penyewaan Sepeda Berdasarkan Jam dan Jenis Hari",
    fontsize=15,
    pad=15
)
plt.xlabel("Jam")
plt.ylabel("Rata-rata Jumlah Penyewaan")
plt.xticks(range(24)) # berfungsi untuk menampilkan semua jam dari 0 hingga 23 pada sumbu x
plt.legend(title="Jenis Hari")
plt.tight_layout()
plt.show()
```

**Insight:**

- Pada hari kerja, terdapat dua periode penyewaan yang tinggi, yaitu sekitar pukul 08.00 dan pukul 17.00–18.00.
- Puncak penyewaan pada hari kerja terjadi pada pukul 17.00.
- Pada hari non-kerja, penyewaan lebih banyak terjadi pada siang hingga sore hari, terutama sekitar pukul 12.00–16.00.
- Pola tersebut menunjukkan bahwa perilaku penyewaan pada hari kerja kemungkinan berkaitan dengan aktivitas berangkat dan pulang kerja, sedangkan pada hari non-kerja lebih banyak berkaitan dengan aktivitas rekreasi atau perjalanan santai.

### Pertanyaan 2:

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=seasonal_rentals,
    x="season_label",
    y="avg_rentals",
    hue="season_label",
    palette="viridis",
    legend=False
)

plt.title(
    "Rata-rata Penyewaan Sepeda Berdasarkan Musim",
    fontsize=15,
    pad=15
)
plt.xlabel("Musim")
plt.ylabel("Rata-rata Jumlah Penyewaan Harian")

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

plt.tight_layout()
plt.show()
```

**Insight:**

- Musim gugur (`Fall`) memiliki rata-rata penyewaan sepeda harian paling tinggi, yaitu sekitar 5.644 penyewaan.
- Musim semi (`Spring`) memiliki rata-rata penyewaan paling rendah, yaitu sekitar 2.604 penyewaan.
- Perbedaan tersebut menunjukkan bahwa musim menjadi salah satu faktor penting yang memengaruhi permintaan penyewaan sepeda.
- Pengelola dapat menyiapkan lebih banyak sepeda dan kapasitas layanan menjelang musim gugur, sedangkan pada musim semi jumlah operasional dapat disesuaikan dengan permintaan yang lebih rendah.

Selain musim, kondisi cuaca juga dianalisis untuk mengetahui pengaruhnya terhadap jumlah penyewaan sepeda harian.

```python
plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=weather_rentals,
    x="weather_label",
    y="avg_rentals",
    hue="weather_label",
    palette="coolwarm",
    legend=False
)

plt.title(
    "Rata-rata Penyewaan Sepeda Berdasarkan Kondisi Cuaca",
    fontsize=15,
    pad=15
)
plt.xlabel("Kondisi Cuaca")
plt.ylabel("Rata-rata Jumlah Penyewaan Harian")

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

plt.tight_layout()
plt.show()
```

**Insight:**

- Kondisi cuaca cerah (`Clear`) memiliki rata-rata penyewaan paling tinggi, yaitu sekitar 4.877 penyewaan per hari.
- Pada kondisi berkabut atau berawan (`Mist/Cloudy`), rata-rata penyewaan turun menjadi sekitar 4.036 penyewaan per hari.
- Pada kondisi hujan atau salju ringan (`Light Rain/Snow`), rata-rata penyewaan turun cukup signifikan menjadi sekitar 1.803 penyewaan per hari.
- Hal ini menunjukkan bahwa cuaca buruk dapat menurunkan permintaan penyewaan sepeda secara signifikan.
